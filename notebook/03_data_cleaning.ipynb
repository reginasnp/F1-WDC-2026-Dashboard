# Step 7: Remove Unnecessary Columns

## 🎯 Objective

Remove columns that are not relevant to the WDC prediction project.

---

## 🇬🇧 English

### Purpose

Some columns contain information that is not required for championship prediction or dashboard visualization. Removing unnecessary columns simplifies the dataset and improves analysis efficiency.

### Expected Output

A cleaner dataset containing only relevant variables for analysis and prediction.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menghapus kolom-kolom yang tidak diperlukan dalam analisis dan prediksi World Drivers' Championship (WDC).

### Output yang Diharapkan

Dataset menjadi lebih ringkas dan hanya berisi variabel yang relevan.

```python
# Columns that are not required for analysis and prediction
columns_to_drop = [
    "BroadcastName",
    "Abbreviation",
    "DriverId",
    "TeamId",
    "TeamColor",
    "HeadshotUrl",
    "CountryCode",
    "Q1",
    "Q2",
    "Q3",
    "Time",
    "FirstName",
    "LastName",
    "ClassifiedPosition",
]

wdc_clean = wdc.drop(
    columns=columns_to_drop,
    errors="ignore"
)

print("✅ Unnecessary columns removed successfully!")
print(f"Original dataset shape : {wdc.shape}")
print(f"Clean dataset shape    : {wdc_clean.shape}")

wdc_clean.head()
```

## 📊 Interpretation

### 🇬🇧 English

Irrelevant columns have been successfully removed from the dataset. The remaining variables are sufficient for exploratory analysis, dashboard development, and WDC prediction.

---

### 🇮🇩 Bahasa Indonesia

Kolom-kolom yang tidak relevan telah berhasil dihapus dari dataset. Variabel yang tersisa sudah cukup untuk proses eksplorasi data, pembuatan dashboard, dan prediksi WDC.

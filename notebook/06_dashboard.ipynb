# Step 5: Driver Standings Dashboard

## 🎯 Objective

Visualize the final driver standings in an interactive dashboard.

---

## 🇬🇧 English

### Purpose

Display the championship points earned by each Formula 1 driver throughout the 2026 season.

### Expected Output

An interactive bar chart showing the current driver standings.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menampilkan klasemen pembalap Formula 1 2026 berdasarkan total poin yang diperoleh.

### Output yang Diharapkan

Grafik batang interaktif yang menunjukkan klasemen pembalap.

```python
driver_standings = (
    wdc.groupby("FullName", as_index=False)["Points"]
    .sum()
    .sort_values("Points", ascending=False)
    .reset_index(drop=True)
)

driver_standings["Rank"] = np.arange(1, len(driver_standings) + 1)

fig_driver = px.bar(
    driver_standings,
    x="FullName",
    y="Points",
    color="Points",
    text="Points",
    color_continuous_scale="Reds",
    title="Formula 1 2026 Driver Championship Standings"
)

fig_driver.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_tickangle=-35,
    coloraxis_showscale=False
)
fig_driver.update_traces(textposition="outside")
fig_driver.show()
fig_driver.write_html(dashboard_path / "driver_standings.html")

display(driver_standings)
```

### Interpretation

This dashboard presents the championship standings by displaying the total points accumulated by each driver. It allows users to quickly identify the leading drivers and compare overall championship performance.

# Step 10: Championship Progression Dashboard

## 🎯 Objective

Visualize how each driver's cumulative championship points changed throughout the season.

---

## 🇬🇧 English

### Purpose

Track the championship battle after every completed Grand Prix.

### Expected Output

An interactive line chart displaying cumulative championship points.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menampilkan perkembangan akumulasi poin setiap pembalap sepanjang musim.

### Output yang Diharapkan

Grafik garis interaktif perkembangan klasemen pembalap.

```python
# Championship Progression by Grand Prix

progression = wdc.copy()

# Define the chronological order of completed Grand Prix
race_order = [
    "Australia",
    "China",
    "Japan",
    "Miami",
    "Canada",
    "Monaco",
    "Spain",
    "Austria",
    "Great Britain",
    "Belgium",
    "Hungary",
    "Netherlands"
]

# Convert GrandPrix into an ordered categorical variable
progression["GrandPrix"] = pd.Categorical(
    progression["GrandPrix"],
    categories=race_order,
    ordered=True
)

# Sort by driver and actual race chronology
progression = progression.sort_values(
    ["FullName", "GrandPrix"]
).copy()

# Calculate cumulative championship points
progression["CumulativePoints"] = (
    progression
    .groupby("FullName", observed=True)["Points"]
    .cumsum()
)

progression.head()
```

```python
fig_progression = px.line(
    progression,
    x="GrandPrix",
    y="CumulativePoints",
    color="FullName",
    markers=True,
    title="Formula 1 2026 Championship Progression"
)

fig_progression.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=650,
    xaxis_title="Grand Prix",
    yaxis_title="Cumulative Championship Points"
)

fig_progression.show()

fig_progression.write_html(
    dashboard_path / "championship_progression.html"
)
```

### Interpretation

The Championship Progression dashboard illustrates how the title battle evolved across the season. It highlights momentum changes, consistency, and the relative performance of each driver after every completed Grand Prix.

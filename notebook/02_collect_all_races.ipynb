# Step 4: Create a Data Collection Function

## 🎯 Objective

Create a reusable Python function to automatically collect race results for any completed Formula 1 Grand Prix.

---

## 🇬🇧 English

### Purpose

Instead of writing the same code for every Grand Prix, we create a reusable function. This improves code readability, reduces duplication, and makes the data collection process scalable.

### Expected Output

A Python function that accepts a Grand Prix name as input and returns the corresponding race results as a DataFrame.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Membuat sebuah fungsi Python yang dapat digunakan kembali untuk mengambil hasil balapan dari Grand Prix mana pun. Dengan cara ini, kita tidak perlu menulis kode yang sama berulang kali.

### Output yang Diharapkan

Sebuah fungsi Python yang menerima nama Grand Prix sebagai input dan mengembalikan hasil balapan dalam bentuk DataFrame.

```python
def collect_race_results(grand_prix):
    """
    Collect official Formula 1 race results for a given Grand Prix.
    """

    print(f"📥 Collecting data for {grand_prix} Grand Prix...")

    # Load race session
    session = fastf1.get_session(2026, grand_prix, "R")
    session.load()

    # Retrieve race results
    results = session.results.copy()

    # Add Grand Prix name as a new column
    results["GrandPrix"] = grand_prix

    return results
```

## 📊 Interpretation

### 🇬🇧 English

A reusable function has been successfully created. Instead of writing the same code repeatedly for every Grand Prix, the function can now collect race results by simply providing the Grand Prix name.

---

### 🇮🇩 Bahasa Indonesia

Sebuah fungsi yang dapat digunakan kembali berhasil dibuat. Dengan fungsi ini, kita tidak perlu lagi menulis kode yang sama untuk setiap Grand Prix. Cukup memberikan nama Grand Prix sebagai input, maka data hasil balapan akan diambil secara otomatis.

# Step 6: Collect All Completed Grand Prix Results

## 🎯 Objective

Automatically collect race results from all completed Formula 1 Grands Prix using a loop.

---

## 🇬🇧 English

### Purpose

Instead of collecting race results manually one by one, we use a loop to automate the process for every completed Grand Prix.

### Expected Output

A combined dataset containing race results from all completed Grands Prix.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Mengotomatisasi proses pengambilan data seluruh Grand Prix yang telah selesai menggunakan perulangan (`for` loop).

### Output yang Diharapkan

Sebuah dataset gabungan yang berisi hasil balapan dari seluruh Grand Prix yang telah selesai.

```python
# Create an empty list to store race results
all_results = []

# Collect data for each completed Grand Prix
for grand_prix in completed_grands_prix:

    race = collect_race_results(grand_prix)

    all_results.append(race)

print(f"\n✅ Successfully collected {len(all_results)} races.")
```

## 📊 Interpretation

### 🇬🇧 English

The automated data collection process completed successfully. Race results from all completed Formula 1 Grands Prix were retrieved and stored in a Python list named `all_results`.

At this stage, each element in the list is a separate DataFrame representing one Grand Prix. These DataFrames will be combined into a single master dataset in the next step.

---

### 🇮🇩 Bahasa Indonesia

Proses pengambilan data secara otomatis berhasil diselesaikan. Hasil balapan dari seluruh Grand Prix Formula 1 yang telah selesai berhasil diambil dan disimpan ke dalam sebuah list Python bernama `all_results`.

Pada tahap ini, setiap elemen di dalam list merupakan sebuah DataFrame yang mewakili satu Grand Prix. Seluruh DataFrame tersebut akan digabungkan menjadi satu dataset utama pada langkah berikutnya.

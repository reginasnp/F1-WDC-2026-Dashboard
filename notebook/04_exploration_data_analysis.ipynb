# Step 4: Driver Championship Standings

## 🎯 Objective

Analyze the current Formula 1 Driver Championship standings based on the total points collected by each driver.

---

## 🇬🇧 English

### Purpose

Calculate the total championship points earned by each driver and visualize the current standings.

### Expected Output

A ranking table and an interactive bar chart displaying the championship standings.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menghitung total poin setiap pembalap dan memvisualisasikan klasemen sementara World Drivers' Championship.

### Output yang Diharapkan

Tabel klasemen pembalap beserta grafik batang interaktif.

```python
driver_standings = (
    wdc
    .groupby("FullName", as_index=False)["Points"]
    .sum()
    .sort_values("Points", ascending=False)
)

driver_standings
```

```python
fig_driver = px.bar(
    driver_standings,
    x="FullName",
    y="Points",
    text="Points",
    title="Formula 1 2026 Driver Championship Standings",
    color="Points",
    color_continuous_scale="Reds"
)

fig_driver.update_traces(
    textposition="outside"
)

fig_driver.update_layout(
    template="plotly_white",
    title_x=0.5,
    title_font_size=22,
    xaxis_title="Driver",
    yaxis_title="Championship Points",
    xaxis_tickangle=-35,
    font=dict(size=13),
    height=600,
    coloraxis_showscale=False
)

fig_driver.show()
```

```python
fig_driver.write_image(
    image_path / "driver_standings.png"
)
```

## 📖 Interpretation

### 🇬🇧 English

The visualization presents the current championship standings after the completed races. Drivers with the highest total points occupy the leading positions and are currently the strongest contenders for the World Drivers' Championship.

---

### 🇮🇩 Bahasa Indonesia

Visualisasi menunjukkan klasemen sementara setelah seluruh balapan yang telah selesai. Pembalap dengan total poin tertinggi berada di posisi teratas dan menjadi kandidat terkuat dalam perebutan gelar juara dunia.

# Step 5: Driver Points Progression

## 🎯 Objective

Calculate the cumulative championship points earned by each driver after every completed Grand Prix.

---

## 🇬🇧 English

### Purpose

Prepare cumulative championship points for each driver that will later be visualized.

### Expected Output

A table containing cumulative points after every race.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menghitung akumulasi poin setiap pembalap setelah setiap Grand Prix.

### Output yang Diharapkan

Tabel perkembangan poin kumulatif setiap pembalap.

```python
wdc.columns.tolist()
```

```python
# Step 5: Driver Points Progression

# Define race order based on completed Grands Prix
race_order = {
    "Australia": 1,
    "China": 2,
    "Japan": 3,
    "Bahrain": 4,
    "Saudi Arabia": 5,
    "Miami": 6,
    "Emilia Romagna": 7,
    "Monaco": 8,
    "Spain": 9,
    "Canada": 10,
    "Austria": 11
}

# Create a copy of the dataset
progression = wdc.copy()

# Create race order column
progression["RaceOrder"] = progression["GrandPrix"].map(race_order)

# Sort data
progression = progression.sort_values(
    ["FullName", "RaceOrder"]
)

# Calculate cumulative points
progression["CumulativePoints"] = (
    progression
    .groupby("FullName")["Points"]
    .cumsum()
)

progression.head(15)
```

## 📖 Interpretation

### 🇬🇧 English

The cumulative points table illustrates how each driver's championship score increases throughout the season. This dataset will be used to create the championship progression visualization.

---

### 🇮🇩 Bahasa Indonesia

Tabel poin kumulatif menunjukkan perkembangan poin setiap pembalap sepanjang musim. Data ini akan digunakan untuk membuat visualisasi perkembangan klasemen pada langkah berikutnya.

# Step 6: Championship Progression Visualization

## 🎯 Objective

Visualize how the championship battle develops throughout the Formula 1 season.

---

## 🇬🇧 English

### Purpose

Display the cumulative championship points earned by each driver after every completed Grand Prix.

# Step 9: Wins by Driver

## 🎯 Objective

Analyze the number of race victories achieved by each driver.

---

## 🇬🇧 English

### Purpose

Calculate the total number of race wins for every driver.

### Expected Output

A ranking table and bar chart of race victories.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menghitung jumlah kemenangan yang diperoleh setiap pembalap.

### Output yang Diharapkan

Tabel dan grafik batang jumlah kemenangan pembalap.

```python
wins = (
    wdc[wdc["Position"] == 1]
    .groupby("FullName", as_index=False)
    .size()
    .rename(columns={"size":"Wins"})
    .sort_values("Wins", ascending=False)
)

wins
```

```python
fig_wins = px.bar(
    wins,
    x="FullName",
    y="Wins",
    text="Wins",
    title="Formula 1 2026 Race Wins",
    color="Wins",
    color_continuous_scale="Reds"
)

fig_wins.update_traces(textposition="outside")

fig_wins.update_layout(
    template="plotly_white",
    title_x=0.5,
    title_font_size=22,
    xaxis_tickangle=-35,
    height=600,
    coloraxis_showscale=False
)

fig_wins.show()
```

```python
fig_wins.write_image(
    image_path / "wins_by_driver.png"
)
```

# Step 11: Average Finishing Position

## 🎯 Objective

Analyze the average finishing position of every driver.

...

```python
average_finish = (
    wdc
    .groupby("FullName", as_index=False)["Position"]
    .mean()
    .rename(columns={"Position":"AverageFinish"})
    .sort_values("AverageFinish")
)

average_finish
```

```python
fig_average_finish = px.bar(
    average_finish,
    x="FullName",
    y="AverageFinish",
    text="AverageFinish",
    title="Formula 1 2026 Average Finishing Position",
    color="AverageFinish",
    color_continuous_scale="Viridis"
)

fig_average_finish.update_traces(
    texttemplate="%{text:.2f}",
    textposition="outside"
)

fig_average_finish.update_layout(
    template="plotly_white",
    title_x=0.5,
    title_font_size=22,
    xaxis_tickangle=-35,
    height=600,
    coloraxis_showscale=False
)

fig_average_finish.show()
```

```python
fig_average_finish.write_image(
    image_path / "average_finish.png"
)
```

## 📖 Interpretation

### 🇬🇧 English

The average finishing position provides an overall measure of each driver's race performance throughout the season. Drivers with lower average finishing positions have demonstrated greater consistency by regularly finishing near the front of the field. This metric complements the championship standings by highlighting performance stability beyond total points alone.

---

### 🇮🇩 Bahasa Indonesia

Rata-rata posisi finis memberikan gambaran umum mengenai performa setiap pembalap sepanjang musim. Pembalap dengan rata-rata posisi finis yang lebih rendah menunjukkan tingkat konsistensi yang lebih baik karena lebih sering menyelesaikan balapan di posisi depan. Metrik ini melengkapi analisis klasemen dengan menyoroti kestabilan performa selain hanya berdasarkan total poin yang diperoleh.

# Step 3: Create Driver Statistics

## 🎯 Objective

Generate aggregated statistics for each Formula 1 driver.

---

## 🇬🇧 English

Summarize each driver's season performance by calculating total races, total championship points, race wins, podium finishes, and average finishing position.

---

## 🇮🇩 Bahasa Indonesia

Membuat ringkasan statistik performa setiap pembalap selama musim berlangsung.

```python
driver_features = (

    wdc

    .groupby("FullName")

    .agg(

        Team=("TeamName","last"),

        TotalRaces=("GrandPrix","count"),

        TotalPoints=("Points","sum"),

        Wins=("Position",lambda x:(x==1).sum()),

        Podiums=("Position",lambda x:(x<=3).sum()),

        AvgFinish=("Position","mean"),

        AvgGrid=("GridPosition","mean")

    )

    .reset_index()

)

driver_features.head()
```

## 📖 Interpretation

The aggregated driver statistics summarize each driver's overall season performance. These variables serve as the foundation for creating more advanced predictive features in the following steps.

# Step 4: Calculate Win Rate

## 🎯 Objective

Calculate the percentage of races won by each driver.

---

## 🇬🇧 English

### Purpose

Measure how frequently each driver converts race entries into victories.

### Expected Output

A new feature called **WinRate** representing the percentage of race wins.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menghitung persentase kemenangan setiap pembalap sepanjang musim.

### Output yang Diharapkan

Sebuah fitur baru bernama **WinRate** yang menunjukkan persentase kemenangan pembalap.

```python
driver_features["WinRate"] = (
    driver_features["Wins"]
    / driver_features["TotalRaces"]
    * 100
).round(2)

driver_features.head()
```

## 📖 Interpretation

### 🇬🇧 English

Win Rate represents the percentage of races won by each driver. A higher win rate indicates greater competitiveness and is one of the strongest indicators of championship success.

---

### 🇮🇩 Bahasa Indonesia

Win Rate menunjukkan persentase kemenangan yang diraih setiap pembalap. Semakin tinggi nilainya, semakin besar tingkat daya saing pembalap dan semakin besar peluangnya dalam perebutan gelar juara dunia.

# Step 5: Calculate Podium Rate

## 🎯 Objective

Calculate the percentage of podium finishes achieved by each driver.

---

## 🇬🇧 English

### Purpose

Measure how consistently a driver finishes within the top three positions.

### Expected Output

A new feature called **PodiumRate**.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menghitung persentase podium yang diperoleh setiap pembalap.

### Output yang Diharapkan

Fitur baru bernama **PodiumRate**.

```python
driver_features["PodiumRate"] = (
    driver_features["Podiums"]
    / driver_features["TotalRaces"]
    * 100
).round(2)

driver_features.head()
```

## 📖 Interpretation

### 🇬🇧 English

Podium Rate measures how consistently a driver finishes in the top three. Drivers with higher podium rates generally demonstrate stronger overall performance and consistency.

---

### 🇮🇩 Bahasa Indonesia

Podium Rate mengukur seberapa konsisten seorang pembalap mampu finis di posisi tiga besar. Nilai yang tinggi menunjukkan performa yang stabil sepanjang musim.

# Step 6: Calculate Points per Race

## 🎯 Objective

Calculate the average championship points earned per race.

---

## 🇬🇧 English

### Purpose

Measure scoring efficiency across the season.

### Expected Output

A new feature named **PointsPerRace**.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menghitung rata-rata poin yang diperoleh setiap pembalap pada setiap balapan.

### Output yang Diharapkan

Fitur baru bernama **PointsPerRace**.

```python
driver_features["PointsPerRace"] = (
    driver_features["TotalPoints"]
    / driver_features["TotalRaces"]
).round(2)

driver_features.head()
```

## 📖 Interpretation

### 🇬🇧 English

Points Per Race reflects a driver's scoring efficiency throughout the season. Drivers with higher values consistently collect championship points.

---

### 🇮🇩 Bahasa Indonesia

Points Per Race menunjukkan efisiensi pembalap dalam mengumpulkan poin setiap balapan. Nilai yang tinggi mencerminkan performa yang lebih konsisten.

# Step 7: Best Finishing Position

## 🎯 Objective

Identify the best finishing position achieved by each driver.

---

## 🇬🇧 English

### Purpose

Determine the driver's highest race finish during the season.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menentukan posisi finis terbaik yang pernah dicapai setiap pembalap.

```python
best_finish = (
    wdc
    .groupby("FullName")["Position"]
    .min()
    .reset_index(name="BestFinish")
)

driver_features = driver_features.merge(
    best_finish,
    on="FullName"
)

driver_features.head()
```

## 📖 Interpretation

### 🇬🇧 English

Best Finish represents the highest race result achieved during the season. It reflects a driver's maximum competitive capability.

---

### 🇮🇩 Bahasa Indonesia

Best Finish menunjukkan hasil finis terbaik yang berhasil diraih pembalap selama musim berlangsung. Fitur ini menggambarkan potensi performa maksimum pembalap.

# Step 8: Worst Finishing Position

## 🎯 Objective

Determine the lowest finishing position achieved by each driver.

...

```python
worst_finish = (
    wdc
    .groupby("FullName")["Position"]
    .max()
    .reset_index(name="WorstFinish")
)

driver_features = driver_features.merge(
    worst_finish,
    on="FullName"
)

driver_features.head()
```

## 📖 Interpretation

Worst Finish captures each driver's poorest finishing result during the season, providing insight into performance variability.

# Step 9: Finish Improvement

## 🎯 Objective

Calculate the average position improvement achieved by each driver during races.

---

## 🇬🇧 English

### Purpose

Measure how many positions a driver gains (or loses) from the starting grid to the finishing position.

### Expected Output

A new feature named **AveragePositionGain**.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menghitung rata-rata perubahan posisi setiap pembalap dari posisi start hingga posisi finis.

### Output yang Diharapkan

Fitur baru bernama **AveragePositionGain**.

```python
# Calculate position gain for each race
wdc["PositionGain"] = (
    wdc["GridPosition"] - wdc["Position"]
)

# Average position gain for each driver
position_gain = (
    wdc
    .groupby("FullName", as_index=False)["PositionGain"]
    .mean()
    .rename(columns={
        "PositionGain": "AveragePositionGain"
    })
)

driver_features = driver_features.merge(
    position_gain,
    on="FullName"
)

driver_features.head()
```

## 📖 Interpretation

### 🇬🇧 English

Average Position Gain measures how effectively a driver improves their race position relative to their starting grid position. Positive values indicate that the driver frequently gains positions during races.

---

### 🇮🇩 Bahasa Indonesia

Average Position Gain menunjukkan kemampuan pembalap dalam memperbaiki posisi dari grid start hingga garis finis. Nilai positif menunjukkan bahwa pembalap sering berhasil menyalip lawan selama balapan.

# Step 10: DNF Rate

## 🎯 Objective

Calculate the percentage of races in which a driver did not finish.

---

## 🇬🇧 English

### Purpose

Measure the reliability and race completion consistency of each driver.

### Expected Output

A new feature called **DNFRate**.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Menghitung persentase balapan yang tidak berhasil diselesaikan oleh setiap pembalap.

### Output yang Diharapkan

Fitur baru bernama **DNFRate**.

```python
# Create DNF indicator
wdc["DNF"] = (
    wdc["Status"] != "Finished"
).astype(int)

# Calculate DNF rate
dnf_rate = (
    wdc
    .groupby("FullName", as_index=False)["DNF"]
    .mean()
)

dnf_rate["DNFRate"] = (
    dnf_rate["DNF"] * 100
).round(2)

dnf_rate = dnf_rate.drop(columns="DNF")

driver_features = driver_features.merge(
    dnf_rate,
    on="FullName"
)

driver_features.head()
```

## 📖 Interpretation

### 🇬🇧 English

DNF Rate represents the percentage of races in which a driver failed to finish. Lower values indicate greater reliability throughout the season.

---

### 🇮🇩 Bahasa Indonesia

DNF Rate menunjukkan persentase balapan yang gagal diselesaikan oleh seorang pembalap. Nilai yang lebih rendah menunjukkan tingkat keandalan dan konsistensi yang lebih baik sepanjang musim.

# Step 2: Configure FastF1 Cache

## 🎯 Objective

Configure a local cache directory to store downloaded Formula 1 data. This helps avoid downloading the same data repeatedly, making future executions much faster.

---

## 🇬🇧 English

### Purpose

FastF1 downloads data from external sources. By enabling a cache, downloaded files are stored locally so they can be reused in future sessions.

### Expected Output

A cache folder is successfully created and FastF1 is configured to use it.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Mengatur folder cache agar data Formula 1 yang sudah diunduh disimpan di komputer. Dengan begitu, saat notebook dijalankan kembali, data tidak perlu diunduh ulang sehingga proses menjadi lebih cepat.

### Output yang Diharapkan

Folder cache berhasil dibuat dan FastF1 siap menggunakan folder tersebut untuk menyimpan data.

```python
from pathlib import Path

# Create cache folder
cache_path = Path("../data/cache")
cache_path.mkdir(parents=True, exist_ok=True)

# Enable FastF1 cache
fastf1.Cache.enable_cache(str(cache_path))

print(f"✅ Cache directory: {cache_path.resolve()}")
print("✅ FastF1 cache enabled successfully!")
```

## 📊 Interpretation

### 🇬🇧 English

The cache directory has been successfully configured. Future requests for the same Formula 1 data will be loaded from the local cache instead of downloading them again.

---

### 🇮🇩 Bahasa Indonesia

Folder cache berhasil dibuat. Ketika data Formula 1 yang sama dibutuhkan lagi, FastF1 akan mengambilnya dari folder cache sehingga proses menjadi lebih cepat dan efisien.

# Step 3: Load Formula 1 Session

## 🎯 Objective

Load an official Formula 1 race session from the 2026 season using FastF1.

---

## 🇬🇧 English

### Purpose

A Formula 1 season consists of multiple Grand Prix weekends, and each weekend contains several sessions such as Practice, Qualifying, Sprint, and Race. In this project, we start by loading one race session to understand the structure of the official Formula 1 data before collecting the entire season.

### Expected Output

The selected Formula 1 race session is successfully loaded into the notebook and ready for further analysis.

---

## 🇮🇩 Bahasa Indonesia

### Tujuan

Musim Formula 1 terdiri dari beberapa Grand Prix, dan setiap Grand Prix memiliki beberapa sesi seperti Practice, Qualifying, Sprint, dan Race. Pada tahap ini, kita akan memuat satu sesi balapan terlebih dahulu untuk memahami struktur data resmi Formula 1 sebelum mengambil seluruh data musim 2026.

### Output yang Diharapkan

Sesi balapan Formula 1 berhasil dimuat ke dalam notebook dan siap digunakan pada tahap analisis berikutnya.

```python
# Load the 2026 Australian Grand Prix Race session
session = fastf1.get_session(
    2026,
    "Australia",
    "R"
)

# Download and load the session data
session.load()

print("✅ Formula 1 session loaded successfully!")
```

---

### 🇮🇩 Bahasa Indonesia

Hasil pemeriksaan missing value menunjukkan bahwa tidak semua kolom memiliki data lengkap. Beberapa variabel seperti **Q1**, **Q2**, dan **Q3** kosong karena dataset yang digunakan berasal dari sesi **Race**, bukan sesi **Qualifying**.
